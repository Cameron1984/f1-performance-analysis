```python
from f1_analysis.data_loader import load_session
import pandas as pd

session = load_session(2024, "Silverstone", "R")

results = session.results
results.iloc[0][["FullName", "Position", "TeamName","Points", "Time"]]
laps = session.laps
```

```python
fastest_laps = laps.groupby('Driver').apply(
    lambda x: x.loc[x['LapTime'].idxmin()] if len(x) > 0 else None
) 

# Sort by lap time
fastest_laps = fastest_laps.sort_values('LapTime')
print(fastest_laps)
```

```python
oscar = laps.pick_drivers("PIA").copy() 
lando = laps.pick_drivers("NOR").copy()
oscar["LapTimeSeconds"] = oscar["LapTime"].dt.total_seconds()

lando["LapTimeSeconds"] = lando["LapTime"].dt.total_seconds()
delta = (
    oscar["LapTimeSeconds"].reset_index(drop=True)
    - lando["LapTimeSeconds"].reset_index(drop=True)
)
for value in delta:
    if value < 0:
        print(f"Norris was {round(value, 3)} faster than Piastri")
    else:
        print(f"Norris was +{round(value, 3)} slower than Piastri")
```

```python
oscar = laps.pick_drivers("PIA").copy() 
lando = laps.pick_drivers("NOR").copy()
oscar["LapTimeSeconds"] = oscar["LapTime"].dt.total_seconds()
lando["LapTimeSeconds"] = lando["LapTime"].dt.total_seconds()
oscar_laps = oscar[["LapNumber", "LapTimeSeconds"]]
lando_laps = lando[["LapNumber", "LapTimeSeconds"]]
compare = oscar_laps.merge(
    lando_laps, 
    on="LapNumber",
    suffixes=("_PIA","_NOR")
)
compare["delta"] = (
    compare["LapTimeSeconds_PIA"]
    - compare["LapTimeSeconds_NOR"] 
)
print(compare["delta"])
```
